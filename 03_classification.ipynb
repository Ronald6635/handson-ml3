### SGD Classifier (SGDClassifier)

**Stochastic Gradient Descent (SGD)** trains linear models by updating weights incrementally using individual samples or small batches. It scales to very large datasets (like MNIST), works with sparse features, and supports online learning via partial_fit.

- Typical uses: linear SVM (loss="hinge") or logistic regression (loss="log_loss")
- Pros: fast, memory-efficient, good for streaming data
- Key hyperparameters:
    - loss (hinge, log_loss), penalty (l2/l1/elasticnet), alpha (regularization)
    - max_iter and tol (early stopping), learning rate params (eta0, learning_rate)
    - class_weight (handle imbalance), random_state (reproducibility)
- Outputs decision_function scores; adjusting the threshold trades precision vs. recall
- Multiclass handled automatically via one-vs-rest (OvR) in scikit-learn

Key Characteristics of Binary Classifiers
Decision Boundary
Binary classifiers create a decision boundary that separates the two classes. The SGD classifier uses a linear decision boundary based on a threshold:

````python
y_scores = sgd_clf.decision_function([some_digit])
threshold = 0
y_prediction = (y_scores > threshold)

print(f"Decision score: {y_scores}")
print(f"Prediction with threshold {threshold}: {y_prediction}")
print(f"Actual label: {y_train_5[0]}")

# y_prediction is a boolean array, e.g., [True]
print(f"The raw prediction is: {y_prediction[0]}")

# Interpret the boolean to get the final class label
if y_prediction[0]:
    final_label = "5"
else:
    final_label = "Not 5"

print(f"\nThe score was {y_scores[0]:.2f}, which is greater than the threshold of {threshold}.")
print(f"Therefore, the final predicted label is: '{final_label}'")
````

### Understanding SGD Classifier Decision Function and Thresholding

This code demonstrates how to use a **Stochastic Gradient Descent (SGD) classifier's decision function** to make predictions with a custom threshold. Let's break down what's happening step by step.

Instead of directly outputting a class label (like "5" or "not-5"), the `decision_function()` method returns **confidence scores**. These scores represent how far a sample is from the decision boundary (hyperplane) that separates different classes. The key insight is that the **sign** of the score determines the predicted class, while the **magnitude** indicates confidence.

In the code, `y_scores = sgd_clf.decision_function([some_digit])` computes the confidence score for a single digit image. The score is a signed distance - positive values suggest the sample belongs to the positive class, while negative values suggest the negative class. The further from zero, the more confident the classifier is about its prediction.

The threshold mechanism `y_prediction = (y_scores > threshold)` allows you to control the **decision boundary**. With `threshold = 0`, you're using the default decision point where the classifier switches between classes. However, you could adjust this threshold based on your needs - for example, setting a higher threshold (like 0.5) would make the classifier more conservative about predicting the positive class, requiring higher confidence.

**Key gotcha**: The decision function returns raw scores, not probabilities. If you need probability estimates, you'd use `predict_proba()` instead. Also, notice that the boolean comparison `(y_scores > threshold)` converts the numerical scores into binary predictions, which is often more useful for final classification decisions than the raw confidence scores themselves.

This approach gives you fine-grained control over the classification decision, which is particularly valuable when dealing with imbalanced datasets or when false positives and false negatives have different costs.

Performance Metrics
Binary classifiers have specific evaluation metrics:
- **Precision**: Of all positive predictions, how many were correct?
- **Recall**: Of all actual positives, how many did we find?
- **F1-Score**: Harmonic mean of precision and recall

## Measuring Accuracy Using Cross-Validation

````python
from sklearn.model_selection import cross_val_score

cross_val_score(sgd_clf, X_train, y_train_5, cv=5, scoring="accuracy")
````

We can also manually implement cross-validation to measure accuracy:

````python
from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone

skfolds = StratifiedKFold(n_splits=5)  # add shuffle=True if the dataset is not
                                       # already shuffled
for train_index, test_index in skfolds.split(X_train, y_train_5):
    clone_clf = clone(sgd_clf)
    X_train_folds = X_train[train_index]
    y_train_folds = y_train_5[train_index]
    X_test_fold = X_train[test_index]
    y_test_fold = y_train_5[test_index]

    clone_clf.fit(X_train_folds, y_train_folds)
    y_pred = clone_clf.predict(X_test_fold)
    n_correct = sum(y_pred == y_test_fold)
    print(n_correct / len(y_pred))
````

*   **`StratifiedKFold(n_splits=5)`**: This is the cross-validation strategy. `cross_val_score` uses `StratifiedKFold` by default for classification tasks when you pass an integer to `cv`.
*   **`for train_index, test_index in skfolds.split(...)`**: Both methods loop through the same train/test splits (folds).
*   **`clone(sgd_clf)`**: `cross_val_score` also clones the estimator for each fold to ensure that it trains a fresh model every time, preventing information from leaking between folds.
*   **`.fit()` and `.predict()`**: Both approaches fit the cloned model on the training fold and make predictions on the test fold.
*   **`n_correct / len(y_pred)`**: This is the manual calculation for the accuracy score. It's the same metric that `scoring="accuracy"` specifies in `cross_val_score`.

In essence, the selected code explicitly shows the step-by-step logic that the `cross_val_score` function abstracts into a single line of code. The purpose of showing the manual implementation in the notebook is to help you understand how cross-validation works under the hood.

## Confusion Matrix

A confusion matrix is a table that summarizes the performance of a classification model by showing the counts of true positive, true negative, false positive, and false negative predictions.

- **True Positives (TP)**: Correctly predicted positive class (5s)
- **True Negatives (TN)**: Correctly predicted negative class (non-5s)
- **False Positives (FP)**: Incorrectly predicted positive class (non-5s predicted as 5s)
- **False Negatives (FN)**: Incorrectly predicted negative class (5s predicted as non-5s)

**F1-Score**: Harmonic mean of precision and recall

### Definitions:
- **Precision**: $\text{Precision} = \frac{TP}{TP + FP}$
- **Recall**: $\text{Recall} = \frac{TP}{TP + FN}$
- **F1-Score**: $\text{F1} = \frac{2}{\frac{1}{\text{Precision}} + \frac{1}{\text{Recall}}} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}} 
= \frac{TP}{TP + \frac{FP + FN}{2}}$

### Meaning:
- **Precision** measures the accuracy of positive predictions. A high precision means that when the model predicts a digit is '5', it is usually correct. This is important in scenarios where false positives are costly (e.g., spam detection).
- **Recall** measures the ability to find all positive instances. A high recall means that the model successfully identifies most of the '5's in the dataset. This is crucial in scenarios where missing positive instances is costly (e.g., disease diagnosis).
- **F1-Score** combines both precision and recall into a single metric. It's particularly useful when you need to consider both false positives and false negatives equally important, or when dealing with imbalanced datasets where accuracy alone might be misleading. A high F1-score indicates that the classifier has both high precision (few false positives) and high recall (few false negatives), making it a more comprehensive metric than accuracy alone.

## Precision/Recall Trade-off

**Precision and Recall Trade-off**

Understanding the precision-recall trade-off is crucial for optimizing binary classifiers. This trade-off arises because increasing precision often comes at the cost of decreasing recall, and vice versa.

**Why the Trade-off Exists:**
- **Higher Threshold → Higher Precision, Lower Recall**: When you increase the decision threshold, the classifier becomes more conservative about predicting the positive class. It will only predict positive when it's very confident, leading to fewer false positives (higher precision) but more false negatives (lower recall).
- **Lower Threshold → Lower Precision, Higher Recall**: With a lower threshold, the classifier becomes more liberal, catching more true positives (higher recall) but also accepting more false positives (lower precision).

**Choosing the Right Balance:**
The optimal balance depends on your specific use case:
- **High Precision Priority**: Use when false positives are costly (e.g., spam detection - you don't want legitimate emails marked as spam)
- **High Recall Priority**: Use when false negatives are costly (e.g., cancer screening - you don't want to miss actual cancer cases)
- **Balanced Approach**: Use F1-score when both precision and recall are equally important

The following visualizations will demonstrate how different thresholds affect this trade-off, helping you make informed decisions about where to set your decision boundary.

**Explanation of cross_val_predict() function**

The `cross_val_predict()` function in scikit-learn is used to generate cross-validated estimates *for each input data point*, i.e., `X_train`. It performs the following steps:
1. **Cross-Validation Splits**: It splits the dataset into `k` folds (subsets) for cross-validation. In each iteration, one fold is used as the validation set while the remaining `k-1` folds are used for training.
2. **Model Training and Prediction**: For each fold, it trains the specified model (e.g., `sgd_clf`) on the training folds and then makes predictions on the validation fold. This process is repeated for all folds.
3. **Collecting Predictions**: It collects the predictions made on each validation fold and combines them to produce a complete set of predictions for the entire dataset. This means that each data point is predicted by a model that was not trained on that specific point, providing an unbiased estimate of the model's performance.

The `y_scores` variable contains the decision function scores for each instance in the training set (`X_train`), obtained through cross-validation. These scores indicate how confident the classifier is about each prediction, with higher positive values suggesting a stronger belief that the instance belongs to the positive class (in this case, the digit '5').

Finally, the code applies standard plot formatting: axis labels, a grid for easier reading, a legend positioned in the lower-left corner, and axis limits set to show the full 0-1 range for both metrics. **Key insight**: This type of visualization is crucial for understanding classifier performance trade-offs. The highlighted point and arrow help viewers understand how adjusting the decision threshold affects the balance between catching more relevant items (recall) versus being more selective about what you classify as positive (precision).

````python
import matplotlib.patches as patches  # extra code – for the curved arrow

plt.figure(figsize=(6, 5))  # extra code – not needed, just formatting

plt.plot(recalls, precisions, linewidth=2, label="Precision/Recall curve")

# extra code – just beautifies and saves Figure 3–6
plt.plot([recalls[idx], recalls[idx]], [0., precisions[idx]], "r:") # repeat x value to make vertical line
plt.plot([0.0, recalls[idx]], [precisions[idx], precisions[idx]], "m:") # repeat y value to make horizontal line
plt.plot([recalls[idx]], [precisions[idx]], "ko",
         label="Point at threshold 3,000") # black dot
# curved arrow
plt.gca().add_patch(patches.FancyArrowPatch(
    (0.79, 0.60), (0.61, 0.78),
    connectionstyle="arc3,rad=.2",
    arrowstyle="Simple, tail_width=1.5, head_width=8, head_length=10",
    color="#444444"))
plt.text(0.56, 0.62, "Higher\nthreshold", color="#333333")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.axis([0, 1, 0, 1])
plt.grid()
plt.legend(loc="lower left")
save_fig("precision_vs_recall_plot")

plt.show()
````

## The ROC Curve
The **ROC (Receiver Operating Characteristic) curve** is a graphical representation of a classifier's performance across different threshold settings. It plots the *True Positive Rate (TPR or Recall)* against the *False Positive Rate (FPR)* at various threshold levels.

````python
from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(y_train_5, y_scores)
````

`fpr, tpr, thresholds = roc_curve(y_train_5, y_scores)
` computes the ROC curve by comparing the true binary labels (`y_train_5`) with the decision scores (`y_scores`) from the classifier. It returns three arrays:
- **fpr**: False Positive Rate which indicates the proportion of actual negatives that were incorrectly classified as positives (false alarms).
- **tpr**: True Positive Rate which indicates the proportion of actual positives that were correctly classified.
- **thresholds**: The thresholds used to compute the FPR and TPR

You can then use these arrays to plot the ROC curve.

````python
# Plot the ROC curve
idx_for_threshold_at_90 = (thresholds <= threshold_for_90_precision).argmax()
tpr_90, fpr_90 = tpr[idx_for_threshold_at_90], fpr[idx_for_threshold_at_90]

plt.figure(figsize=(6, 5))  # extra code – not needed, just formatting
plt.plot(fpr, tpr, linewidth=2, label="ROC curve")
plt.plot([0, 1], [0, 1], 'k:', label="Random classifier's ROC curve")
plt.plot([fpr_90], [tpr_90], "kx", label="Threshold for 90% precision")

# extra code – just beautifies and saves Figure 3–7
plt.gca().add_patch(patches.FancyArrowPatch(
    (0.20, 0.89), (0.07, 0.70),
    connectionstyle="arc3,rad=.4",
    arrowstyle="Simple, tail_width=1.5, head_width=8, head_length=10",
    color="#444444"))
plt.text(0.12, 0.71, "Higher\nthreshold", color="#333333")
plt.xlabel('False Positive Rate (Fall-Out)')
plt.ylabel('True Positive Rate (Recall)')
plt.grid()
plt.axis([0, 1, 0, 1])
plt.legend(loc="lower right", fontsize=13)
save_fig("roc_curve_plot")

plt.show()
````

# Multilabel Classification

## Understanding Multilabel Classification

Multilabel classification is a variant of classification where **each instance can belong to multiple classes simultaneously**, unlike multiclass classification where each instance belongs to exactly one class.

**Key Characteristics:**
- **Multiple Labels per Instance**: A single data point can have several target labels
- **Independent Predictions**: Each label is typically predicted independently
- **Real-world Applications**: Email categorization (spam + urgent + work), image tagging (person + outdoor + daytime), medical diagnosis (multiple conditions)

**Example Scenarios:**
- **Document Classification**: A news article could be tagged as both "Politics" and "Economics"
- **Image Recognition**: A photo might contain both "person" and "car" and "outdoor"
- **Gene Classification**: A gene might be associated with multiple biological functions

**Implementation Approaches:**
1. **Binary Relevance**: Train separate binary classifiers for each label
2. **Classifier Chains**: Account for label dependencies by chaining classifiers
3. **Label Powerset**: Treat each unique combination of labels as a single class

**Evaluation Metrics:**
- **Exact Match Ratio**: Percentage of instances where all labels are predicted correctly
- **Hamming Loss**: Average fraction of incorrect labels per instance
- **F1-score variants**: Micro-averaged, macro-averaged, or per-label F1 scores

The following code demonstrates how to implement and evaluate multilabel classification using scikit-learn's approaches.

````python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

# Multilabel classification
# Classify an image as one of two classes: large (7, 8, 9) vs small (0-6)
# and odd vs even
y_train_large = (y_train >= '7')
y_train_odd = (y_train.astype('int8') % 2 == 1)
y_multilabel = np.c_[y_train_large, y_train_odd] # Combine the two labels into a 2D array
print(f'Shape of y_multilabel: {y_multilabel.shape}') # extra code – just to show the shape of y_multilabel

knn_clf = KNeighborsClassifier()
knn_clf.fit(X_train, y_multilabel)
````

## Evaluating Multilabel Classification Performance

The following code evaluates the multilabel classifier's performance using cross-validation:

**Key Components:**

**Cross-Validation Prediction:**
- `cross_val_predict()` generates predictions for each sample using 3-fold cross-validation
- Each sample is predicted by a model that wasn't trained on that specific sample
- Returns predictions for both labels (large/small and odd/even) for all training instances

**F1-Score with Macro Averaging:**
- `average="macro"` calculates the F1-score for each label independently, then takes their average
- This gives equal weight to both the "large vs small" and "odd vs even" classification tasks
- Alternative averaging methods:
    - `"micro"`: Calculate metrics globally by counting total true positives, false negatives, etc.
    - `"weighted"`: Average weighted by the number of true instances for each label
    - `None`: Return F1-scores for each label separately

**Interpretation:**
The resulting score represents the overall multilabel classification performance, balancing both the digit size classification (≥7 vs <7) and parity classification (odd vs even) tasks equally.

# Multioutput Classification

**Multioutput classification** is a generalization of multilabel classification where each label can be **multiclass** (not just binary). This means each output can have more than two possible values.

**Examples:**
- **Face Recognition**: Predicting both the person's identity (Alice, Bob, Charlie, etc.) AND their emotion (happy, sad, angry, etc.)
- **Image Segmentation**: Predicting multiple attributes for each pixel (object type, color category, lighting condition)
- **Weather Prediction**: Predicting both temperature range (cold, mild, warm, hot) AND precipitation type (none, rain, snow, sleet)

**Key Difference from Multilabel:**
- **Multilabel**: Multiple binary outputs (yes/no for each label)
- **Multioutput**: Multiple categorical outputs (each can have 2+ categories)

**Classifier Chains** is a popular approach for both multilabel and multioutput classification that accounts for **label dependencies**:

1. **Sequential Training**: Trains classifiers in a chain, where each classifier predicts one label
2. **Feature Augmentation**: Each classifier uses original features PLUS predictions from previous classifiers in the chain
3. **Dependency Modeling**: Later classifiers can learn relationships between labels
4. **Order Matters**: The sequence of labels affects performance, so order selection is important

**Advantages:**
- Models label correlations (unlike independent binary classifiers)
- Often improves performance when labels are dependent
- Flexible framework that works with any base classifier

**Disadvantages:**
- More complex than independent classifiers
- Prediction errors can propagate through the chain
- Requires careful consideration of label ordering

## Multioutput Classification: Noise Reduction Example

The code above demonstrates a practical **multioutput classification** scenario where we train a model to remove noise from digit images. Here's what each step accomplishes:

**Adding Synthetic Noise:**
```python
noise = np.random.randint(0, 100, (len(X_train), 784))
X_train_mod = X_train + noise
```
- Generates random integer noise values between 0-99 for each pixel
- Adds this noise to the original clean training images
- Creates a "corrupted" version of the MNIST dataset

**Setting Up the Learning Task:**
```python

y_test_mod = X_test
```
- Uses the **original clean images as targets** (`y_train_mod`)
- The model learns to map from **noisy inputs** (`X_train_mod`) to **clean outputs** (`y_train_mod`)

**Multioutput Nature:**
- Each pixel position (784 total) becomes a separate output
- Each output predicts the clean pixel intensity (0-255)
- This creates a **784-class multioutput problem** where each output can have 256 possible values

**Real-World Applications:**
- **Image denoising**: Removing camera sensor noise, compression artifacts
- **Image restoration**: Fixing damaged or degraded historical photographs  
- **Super-resolution**: Enhancing low-resolution images to higher resolution
- **Medical imaging**: Improving clarity of X-rays, MRIs, or CT scans

This approach demonstrates how multioutput classification can solve complex regression-like problems by treating each continuous output as a discrete classification task across pixel intensities.

This function creates shifted versions of MNIST images for data augmentation:

- **Reshapes the flattened image** (784,) back to 2D format (28, 28) for processing
- **Uses scipy's shift function** with `[dy, dx]` parameter order (vertical, horizontal)
- **Sets `cval=0`** to fill empty pixels with black (0 intensity) after shifting
- **Uses `mode="constant"`** to avoid wrapping around pixels
- **Returns flattened format** (784,) to maintain consistency with original data format

The function will be used to create four additional training samples per original image by shifting in each direction (up, down, left, right), effectively expanding the training set by 5x.

````python
def shift_image(image, dx, dy):
    image = image.reshape((28, 28))
    shifted_image = shift(image, [dy, dx], cval=0, mode="constant")
    return shifted_image.reshape([-1])
````

Okay, let's start writing the preprocessing functions. First, we will need a function to convert HTML to plain text. Arguably the best way to do this would be to use the great [BeautifulSoup](https://www.crummy.com/software/BeautifulSoup/) library, but I would like to avoid adding another dependency to this project, so let's hack a quick & dirty solution using regular expressions (at the risk of [un̨ho͞ly radiańcé destro҉ying all enli̍̈́̂̈́ghtenment](https://stackoverflow.com/a/1732454/38626)). The following function first drops the `<head>` section, then converts all `<a>` tags to the word HYPERLINK, then it gets rid of all HTML tags, leaving only the plain text. For readability, it also replaces multiple newlines with single newlines, and finally it unescapes html entities (such as `&gt;` or `&nbsp;`):

````python
import re
from html import unescape

def html_to_plain_text(html):
    text = re.sub('<head.*?>.*?</head>', '', html, flags=re.M | re.S | re.I)
    text = re.sub('<a\s.*?>', ' HYPERLINK ', text, flags=re.M | re.S | re.I)
    text = re.sub('<.*?>', '', text, flags=re.M | re.S)
    text = re.sub(r'(\s*\n)+', '\n', text, flags=re.M | re.S)
    return unescape(text)
````

Great! Now let's write a function that takes an email as input and returns its content as plain text, whatever its format is:

````python
def email_to_text(email):
    html = None
    for part in email.walk():
        ctype = part.get_content_type()
        if not ctype in ("text/plain", "text/html"):
            continue
        try:
            content = part.get_content()
        except: # in case of encoding issues
            content = str(part.get_payload())
        if ctype == "text/plain":
            return content
        else:
            html = content
    if html:
        return html_to_plain_text(html)
````

We are ready to put all this together into a transformer that we will use to convert emails to word counters. Note that we split sentences into words using Python's `split()` method, which uses whitespaces for word boundaries. This works for many written languages, but not all. For example, Chinese and Japanese scripts generally don't use spaces between words, and Vietnamese often uses spaces even between syllables. It's okay in this exercise, because the dataset is (mostly) in English.

````python
from sklearn.base import BaseEstimator, TransformerMixin

class EmailToWordCounterTransformer(BaseEstimator, TransformerMixin):
    def __init__(self, strip_headers=True, lower_case=True,
                 remove_punctuation=True, replace_urls=True,
                 replace_numbers=True, stemming=True):
        self.strip_headers = strip_headers
        self.lower_case = lower_case
        self.remove_punctuation = remove_punctuation
        self.replace_urls = replace_urls
        self.replace_numbers = replace_numbers
        self.stemming = stemming
    def fit(self, X, y=None):
        return self
    def transform(self, X, y=None):
        X_transformed = []
        for email in X:
            text = email_to_text(email) or ""
            if self.lower_case:
                text = text.lower()
            if self.replace_urls and url_extractor is not None:
                urls = list(set(url_extractor.find_urls(text)))
                urls.sort(key=lambda url: len(url), reverse=True)
                for url in urls:
                    text = text.replace(url, " URL ")
            if self.replace_numbers:
                text = re.sub(r'\d+(?:\.\d*)?(?:[eE][+-]?\d+)?', 'NUMBER', text)
            if self.remove_punctuation:
                text = re.sub(r'\W+', ' ', text, flags=re.M)
            word_counts = Counter(text.split())
            if self.stemming and stemmer is not None:
                stemmed_word_counts = Counter()
                for word, count in word_counts.items():
                    stemmed_word = stemmer.stem(word)
                    stemmed_word_counts[stemmed_word] += count
                word_counts = stemmed_word_counts
            X_transformed.append(word_counts)
        return np.array(X_transformed)
````

Now we have the word counts, and we need to convert them to vectors. For this, we will build another transformer whose `fit()` method will build the vocabulary (an ordered list of the most common words) and whose `transform()` method will use the vocabulary to convert word counts to vectors. The output is a sparse matrix.

````python
from scipy.sparse import csr_matrix

class WordCounterToVectorTransformer(BaseEstimator, TransformerMixin):
    def __init__(self, vocabulary_size=1000):
        self.vocabulary_size = vocabulary_size
    def fit(self, X, y=None):
        total_count = Counter()
        for word_count in X:
            for word, count in word_count.items():
                total_count[word] += min(count, 10)
        most_common = total_count.most_common()[:self.vocabulary_size]
        self.vocabulary_ = {word: index + 1
                            for index, (word, count) in enumerate(most_common)}
        return self
    def transform(self, X, y=None):
        rows = []
        cols = []
        data = []
        for row, word_count in enumerate(X):
            for word, count in word_count.items():
                rows.append(row)
                cols.append(self.vocabulary_.get(word, 0))
                data.append(count)
        return csr_matrix((data, (rows, cols)),
                          shape=(len(X), self.vocabulary_size + 1))
````
