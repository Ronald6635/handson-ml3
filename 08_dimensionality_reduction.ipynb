Let's generate a small 3D dataset. It's an oval shape, rotated in 3D space, with points distributed unevenly, and with quite a lot of noise:

````python
# extra code

import numpy as np
from scipy.spatial.transform import Rotation

m = 60
X = np.zeros((m, 3))  # initialize 3D dataset
np.random.seed(42)
angles = (np.random.rand(m) ** 3 + 0.5) * 2 * np.pi  # uneven distribution
X[:, 0], X[:, 1] = np.cos(angles), np.sin(angles) * 0.5  # oval
X += 0.28 * np.random.randn(m, 3)  # add more noise
X = Rotation.from_rotvec([np.pi / 29, -np.pi / 20, np.pi / 4]).apply(X)
X += [0.2, 0, 0.2]  # shift a bit
````

Plot the 3D dataset, with the projection plane.

````python
# extra code – this cell generates and saves Figure 8–2

import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X2D = pca.fit_transform(X)  # dataset reduced to 2D
X3D_inv = pca.inverse_transform(X2D)  # 3D position of the projected samples
X_centered = X - X.mean(axis=0)
U, s, Vt = np.linalg.svd(X_centered)

axes = [-1.4, 1.4, -1.4, 1.4, -1.1, 1.1]
x1, x2 = np.meshgrid(np.linspace(axes[0], axes[1], 10),
                     np.linspace(axes[2], axes[3], 10))
w1, w2 = np.linalg.solve(Vt[:2, :2], Vt[:2, 2])  # projection plane coefs
z = w1 * (x1 - pca.mean_[0]) + w2 * (x2 - pca.mean_[1]) - pca.mean_[2]  # plane
X3D_above = X[X[:, 2] >= X3D_inv[:, 2]]  # samples above plane
X3D_below = X[X[:, 2] < X3D_inv[:, 2]]  # samples below plane

fig = plt.figure(figsize=(9, 9))
ax = fig.add_subplot(111, projection="3d")

# plot samples and projection lines below plane first
ax.plot(X3D_below[:, 0], X3D_below[:, 1], X3D_below[:, 2], "ro", alpha=0.3)
for i in range(m):
    if X[i, 2] < X3D_inv[i, 2]:
        ax.plot([X[i][0], X3D_inv[i][0]],
                [X[i][1], X3D_inv[i][1]],
                [X[i][2], X3D_inv[i][2]], ":", color="#F88")

ax.plot_surface(x1, x2, z, alpha=0.1, color="b")  # projection plane
ax.plot(X3D_inv[:, 0], X3D_inv[:, 1], X3D_inv[:, 2], "b+")  # projected samples
ax.plot(X3D_inv[:, 0], X3D_inv[:, 1], X3D_inv[:, 2], "b.")

# now plot projection lines and samples above plane
for i in range(m):
    if X[i, 2] >= X3D_inv[i, 2]:
        ax.plot([X[i][0], X3D_inv[i][0]],
                [X[i][1], X3D_inv[i][1]],
                [X[i][2], X3D_inv[i][2]], "r--")

ax.plot(X3D_above[:, 0], X3D_above[:, 1], X3D_above[:, 2], "ro")

def set_xyz_axes(ax, axes):
    ax.xaxis.set_rotate_label(False)
    ax.yaxis.set_rotate_label(False)
    ax.zaxis.set_rotate_label(False)
    ax.set_xlabel("$x_1$", labelpad=8, rotation=0)
    ax.set_ylabel("$x_2$", labelpad=8, rotation=0)
    ax.set_zlabel("$x_3$", labelpad=8, rotation=0)
    ax.set_xlim(axes[0:2])
    ax.set_ylim(axes[2:4])
    ax.set_zlim(axes[4:6])

set_xyz_axes(ax, axes)
ax.set_zticks([-1, -0.5, 0, 0.5, 1])

save_fig("dataset_3d_plot", tight_layout=False)
plt.show()
````

This code generates **Figure 8–2**, which visualizes a small 3D dataset (an oval shape rotated in 3D space with added noise) and its projection onto a 2D plane using PCA. The plot shows:

- **Projection plane**: A semi-transparent blue surface representing the best-fit 2D subspace found by PCA.
- **Projected points**: Blue crosses and dots on the plane, indicating where the 3D points map in 2D space.

The code uses `sklearn.decomposition.PCA` to reduce the dataset to 2D, computes the projection plane coefficients via SVD, and employs Matplotlib's 3D plotting to render the scene. ## Principal Components
Principal components are the directions in the feature space along which the data varies the most. In PCA, these are found by performing singular value decomposition (SVD) on the mean-centered data matrix. The first principal component captures the maximum variance, the second captures the next highest, and so on. This allows us to reduce dimensionality by projecting the data onto the top principal components while preserving as much variance as possible. Let's explore how to compute them manually and using Scikit-Learn.

````python
import numpy as np

# X = [...]  # the small 3D dataset was created earlier in this notebook
X_centered = X - X.mean(axis=0)
U, s, Vt = np.linalg.svd(X_centered)
c1 = Vt[0]
c2 = Vt[1]
````

````python
# extra code – plot X_centered with c1 and c2
fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111, projection="3d")
ax.plot(X_centered[:, 0], X_centered[:, 1], X_centered[:, 2], "ro")
# Plot principal components as arrows from origin
ax.quiver([0], [0], [0], [c1[0] * 0.25 * s[0]], [c1[1] * 0.25 * s[0]], [c1[2] * 0.25 * s[0]], color='b', linewidth=4)
ax.quiver([0], [0], [0], [c2[0] * 0.25 * s[1]], [c2[1] * 0.25 * s[1]], [c2[2] * 0.25 * s[1]], color='b', linewidth=1)
plt.show()
````

Note: in principle, the SVD factorization algorithm returns three matrices, **U**, **Σ** and **V**, such that **X** = **UΣV**<sup>⊺</sup>, where **U** is an _m_ × _m_ matrix, **Σ** is an _m_ × _n_ matrix, and **V** is an _n_ × _n_ matrix. But the `svd()` function returns **U**, **s** and **V**<sup>⊺</sup> instead. **s** is the vector containing all the values on the main diagonal of the top _n_ rows of **Σ**. Since **Σ** is full of zeros elsewhere, your can easily reconstruct it from **s**, like this:

````python
# extra code – shows how to construct Σ from s
m, n = X.shape
Σ = np.zeros_like(X_centered)
Σ[:n, :n] = np.diag(s)
assert np.allclose(X_centered, U @ Σ @ Vt), "SVD reconstruction error"
````

### Mathematical Explanation of SVD

Singular Value Decomposition (SVD) is a fundamental matrix factorization technique in linear algebra that decomposes a matrix into three constituent matrices, revealing important properties about the data. It is widely used in dimensionality reduction, data compression, and noise reduction, as seen in Principal Component Analysis (PCA).

#### Definition and Formula
For a real-valued matrix **X** of size _m_ × _n_ (where _m_ is the number of samples and _n_ is the number of features), SVD decomposes **X** as:

**X** = **U Σ V**<sup>⊺</sup>

Where:
- **U** is an _m_ × _m_ orthogonal matrix (_U_<sup>⊺</sup>**U** = **I**), whose columns are the **left singular vectors**.
- **Σ** is an _m_ × _n_ diagonal matrix containing the **singular values** (non-negative real numbers) on the diagonal, sorted in descending order, indicating the "strength" or importance of each component. All off-diagonal elements are zero.
- **V**<sup>⊺</sup> is an _n_ × _n_ orthogonal matrix (_V_<sup>⊺</sup>**V** = **I**), whose rows are the **right singular vectors** (often denoted as **V**<sup>⊺</sup> for the transpose), representing the bases for the row space.
    - The columns of **V** correspond to the principal components in PCA.
    $$ \mathbf{V} = [\mathbf{c_1}, \mathbf{c_2}, \ldots, \mathbf{c_n}] $$
    - Each column $\mathbf{c_i}$ is a principal component direction (an eigenvector of the covariance matrix of the data **X**), and together they form an orthonormal basis for the feature space.In PCA, these principal components are used to project the data into a lower-dimensional space while preserving as much variance as possible.

In NumPy's `np.linalg.svd()`, the function returns **U**, **s** (a 1D array of singular values), and **V**<sup>⊺</sup> (the transpose of **V**). To reconstruct **Σ**, we can create a diagonal matrix from **s**:

```python
Σ = np.zeros_like(X_centered)
Σ[:n, :n] = np.diag(s)
```

This ensures **X** can be reconstructed as **U @ Σ @ V**<sup>⊺</sup>, and the reconstruction error is minimal.

#### Interpretation in Dimensionality Reduction
- The singular values in **Σ** represent the importance of each dimension: larger values indicate directions with more variance.
- The first _k_ columns of **U** and rows of **V**<sup>⊺</sup> correspond to the top _k_ principal components.
- By retaining only the top _k_ singular values and vectors, we approximate **X** with lower rank, reducing dimensionality while minimizing the Frobenius norm of the error.

In the context of this notebook, SVD is applied to the mean-centered data `X_centered` (derived from `X`, a 60×3 array of 3D points), yielding:
- **U**: Shape (60, 60), left singular vectors.
- **s**: Array of singular values, e.g., [5.93449971, 2.65661127, 2.04831319], indicating the scale of variance along each principal direction.
- **V**<sup>⊺</sup>: Shape (3, 3), right singular vectors, from which principal components `c1`, `c2`, and `c3` are extracted.

This decomposition underpins PCA, as the principal components are the columns of **V**, and the explained variance relates directly to the squared singular values. For example, projecting onto the first two components reduces the 3D data to 2D while preserving ~91% of the variance.

**What SVD Does (The Intuition)**

Singular Value Decomposition (SVD) decomposes a matrix **X** (representing the dataset) into three matrices: **X** = **U Σ V**<sup>⊺</sup>, where:

- **V**<sup>⊺</sup> rotates the original coordinate system to align with the directions of maximum variance (the principal components).
    - The input space is rotated by Vᵀ.
- **Σ** scales the data along these new axes by the singular values, which quantify the amount of variance in each direction.
    - The data is stretched or shrunk along the axes (defined by Vᵀ) according to the singular values in Σ.
- **U** rotates the scaled data into the output space.

This sequence of transformations captures the essential structure of the data, making SVD a powerful tool for dimensionality reduction, noise reduction, and data analysis by retaining only the most significant components.

### Mathematical Explanation of Projecting Down to d Dimensions

In the context of Principal Component Analysis (PCA), projecting the data onto a lower-dimensional subspace is a key step for dimensionality reduction. This process leverages the Singular Value Decomposition (SVD) to transform the original high-dimensional data into a reduced space while preserving as much variance as possible.

#### Projection Formula
Given the mean-centered data matrix **X** (of shape _m_ × _n_, where _m_ is the number of samples and _n_ is the number of features), SVD decomposes **X** as:

**X** = **U Σ V**<sup>⊺</sup>

To project the data onto the first _d_ principal components (where _d_ < _n_), we use the matrix **W**<sub>d</sub>, which consists of the first _d_ principal components as its columns. The projection is computed as:

**X**<sub>projected</sub> = **X**<sub>centered</sub> **W**<sub>d</sub>

Where:
- **X**<sub>centered</sub> is the mean-centered data matrix.
- **W**<sub>d</sub> is an _n_ × _d_ matrix whose columns are the first _d_ principal components (derived from the first _d_ rows of **V**<sup>⊺</sup>, transposed for column-major storage to ensure columns represent the PCs).

This projection minimizes the reconstruction error in the Frobenius norm, ensuring that the reduced data retains the most significant variance.

#### Interpretation
- The principal components in **W**<sub>d</sub> represent the directions of maximum variance in the data.
- The explained variance for each component is given by the squared singular values: variance explained by component _i_ = σ<sub>i</sub><sup>2</sup> / ∑ σ<sub>j</sub><sup>2</sup>, where σ<sub>i</sub> is the singular value corresponding to the _i_-th principal component.
- By selecting _d_ such that the cumulative explained variance reaches a threshold (e.g., 95%), we balance dimensionality reduction with information preservation.

In this notebook, for the 3D dataset `X` (60 samples, 3 features), projecting onto the first 2 principal components reduces it to 2D, explaining approximately 91% of the variance. This is visualized in the scatter plot of `X2D`, where the axes correspond to the principal components.

For larger datasets like MNIST (784 features), choosing _d_ = 154 retains 95% variance, compressing the data significantly while maintaining essential structure for tasks like classification. This projection is the foundation of PCA's effectiveness in high-dimensional data analysis.

````python
from sklearn.datasets import fetch_openml
from sklearn.decomposition import PCA
import numpy as np

mnist = fetch_openml('mnist_784', as_frame=False, parser="auto")
X_train, y_train = mnist.data[:60_000], mnist.target[:60_000]
X_test, y_test = mnist.data[60_000:], mnist.target[60_000:]
````

````python
# PCA to find d for 95% variance retention
pca = PCA()
pca.fit(X_train)
cumsum = np.cumsum(pca.explained_variance_ratio_)
d = np.argmax(cumsum >= 0.95) + 1  # d equals 154
````

## Incremental PCA

Incremental PCA is a variant of PCA that allows you to process large datasets in smaller batches, making it suitable for datasets that do not fit into memory. Instead of computing the principal components on the entire dataset at once, Incremental PCA updates the components incrementally as it processes each batch of data. This approach is particularly useful for streaming data or when working with very large datasets. Here's an example of how to use Incremental PCA with the MNIST dataset:

````python
from sklearn.decomposition import IncrementalPCA

n_batches = 100 # number of mini-batches
inc_pca = IncrementalPCA(n_components=154)
for X_batch in np.array_split(X_train, n_batches): # split into mini-batches
    inc_pca.partial_fit(X_batch) # incrementally fit on each mini-batch

X_reduced = inc_pca.transform(X_train)
````

````python
inc_pca.explained_variance_ratio_.sum()
````

````python
cumsum_rnd_pca = np.cumsum(rnd_pca.explained_variance_ratio_)
cumsum_inc_pca = np.cumsum(inc_pca.explained_variance_ratio_)
plt.figure(figsize=(16, 6))
plt.subplot(121)
plt.plot(cumsum_rnd_pca, linewidth=3, label="Randomized PCA")
plt.plot(cumsum_inc_pca, linewidth=3, label="Incremental PCA")
plt.legend()   
plt.grid(True) 
plt.subplot(122)
plt.plot(np.linspace(len(cumsum_rnd_pca)-9, len(cumsum_rnd_pca), 10), cumsum_rnd_pca[-10:], linewidth=3, label="Randomized PCA")
plt.plot(np.linspace(len(cumsum_inc_pca)-9, len(cumsum_inc_pca), 10), cumsum_inc_pca[-10:], linewidth=3, label="Incremental PCA")
plt.legend()   
plt.xlim(len(cumsum_rnd_pca)-9, len(cumsum_rnd_pca))
plt.grid(True)
plt.show()
````

# Random Projection

Random Projection is a dimensionality reduction technique that projects high-dimensional data onto a lower-dimensional subspace using a random matrix. Unlike PCA, which seeks to preserve variance along principal components, Random Projection aims to *preserve pairwise distances between points with high probability*, as guaranteed by the Johnson-Lindenstrauss lemma. This makes it computationally efficient and suitable for very high-dimensional datasets, often used as a preprocessing step for other algorithms. It can be implemented using *Gaussian or sparse random matrices*, and while it may not capture as much structure as PCA, it provides a fast approximation for tasks like clustering or classification.

**Warning**: this sections will use close to 2.5 GB of RAM. If your computer runs out of memory, just reduce _m_ and _n_:

````python
from sklearn.random_projection import johnson_lindenstrauss_min_dim

m, ε = 5_000, 0.1
d = johnson_lindenstrauss_min_dim(m, eps=ε)
d
````

This code calculates the minimum number of dimensions required for random projection to preserve pairwise distances in a dataset of 5,000 samples with an error tolerance of 0.1, using the Johnson-Lindenstrauss lemma. The `johnson_lindenstrauss_min_dim` function from Scikit-Learn determines this dimension `d`, ensuring that distances are preserved with high probability during dimensionality reduction. This is a precursor to applying random projection techniques for efficient data compression.

Next, manually compute the minimum dimension `d` required for random projection using the Johnson-Lindenstrauss lemma formula: `d = 4 * log(m) / (ε² / 2 - ε³ / 3)`, where `m` is the number of instances (5,000) and `ε` is the error tolerance (0.1). This approximation matches the result from `johnson_lindenstrauss_min_dim`, providing an analytical way to *estimate the dimensionality needed to preserve pairwise distances with high probability* during random projection. This is useful for understanding the theoretical basis and for quick calculations without relying on the Scikit-Learn function.

````python
# extra code – show the equation computed by johnson_lindenstrauss_min_dim
d = int(4 * np.log(m) / (ε ** 2 / 2 - ε ** 3 / 3))
d
````

Generate a projection matrix `P` of shape (d, n) using a Gaussian distribution to reduce the dimensionality of a dataset. The random projection matrix is created with entries drawn from a standard normal distribution and scaled by `1/sqrt(d)` to ensure that the projected data maintains the original scale along the projection axis (i.e., the length of the projected vectors). This matrix can then be used to project high-dimensional data into a lower-dimensional space while approximately preserving pairwise distances, as per the Johnson-Lindenstrauss lemma.

````python
n = 20_000
np.random.seed(42)
P = np.random.randn(d, n) / np.sqrt(d)  # std dev = square root of variance

X = np.random.randn(m, n)  # generate a fake dataset
X_reduced = X @ P.T
````

This code implements a manual random projection to reduce the dimensionality of a synthetic high-dimensional dataset, demonstrating the Johnson-Lindenstrauss lemma in action. It generates a random projection matrix `P` and applies it to project the data from 20,000 dimensions down to approximately 7,300 dimensions (based on the previously computed `d`), preserving pairwise distances with high probability.

- **`n = 20_000`**: Sets the original dimensionality (number of features) for the dataset.
- **`np.random.seed(42)`**: Ensures reproducibility of the random matrix generation.
- **`P = np.random.randn(d, n) / np.sqrt(d)`**: Creates a **Gaussian random projection matrix** `P` of shape (`d`, `n`), where each entry is drawn from a normal distribution with mean 0 and variance 1/`d`. This normalization ensures the projection preserves distances approximately, as per the lemma.
- **`X = np.random.randn(m, n)`**: Generates a fake dataset `X` with `m` samples (5,000) and `n` features (20,000), simulating high-dimensional data.
- **`X_reduced = X @ P.T`**: Projects the dataset `X` onto the lower-dimensional subspace by multiplying it with the transpose of `P`, resulting in `X_reduced` of shape (5,000, `d`). This reduces the data while aiming to maintain relative distances between points.

This approach is computationally efficient for very high-dimensional data and serves as an alternative to PCA for tasks like clustering or preprocessing, though it may not capture as much variance structure. The resulting `X_reduced` can be used for further analysis or visualization.

### Explanation of Generating a Random Projection Matrix

Random projection is a dimensionality reduction technique that uses a random matrix to project high-dimensional data into a lower-dimensional space, often as a faster alternative to methods like PCA. The active selection describes creating a projection matrix $\mathbf{P}$ with shape $(d, n)$, where $n$ is the original dimensionality (e.g., 20,000 features) and $d$ is the target lower dimensionality (e.g., 7,300). This matrix is generated by drawing entries from a standard normal distribution (mean 0, variance 1) and then scaling each entry by $1 / \sqrt{d}$. The scaling is crucial because it normalizes the matrix to preserve the expected Euclidean norm of the projected vectors, ensuring that distances in the original space are approximately maintained in the reduced space.

Mathematically, for a data point $\mathbf{x} \in \mathbb{R}^n$, the projection is $\mathbf{x}' = \mathbf{P} \mathbf{x}$, resulting in $\mathbf{x}' \in \mathbb{R}^d$. Without scaling, the variance of each component in $\mathbf{x}'$ would be $n$ times larger than in $\mathbf{x}$, distorting scales. By dividing by $\sqrt{d}$, the expected squared norm of $\mathbf{x}'$ matches that of $\mathbf{x}$, as $\mathbb{E}[\|\mathbf{P} \mathbf{x}\|^2] = \|\mathbf{x}\|^2$. This ties into the Johnson-Lindenstrauss lemma, which guarantees that with high probability, pairwise distances between points are preserved up to a factor of $1 \pm \epsilon$ (for small $\epsilon$), provided $d$ is chosen appropriately (e.g., via the lemma's formula based on the number of points and error tolerance).

In practice, this approach is computationally efficient—matrix generation and projection are $O(n \cdot d)$—and works well for high-dimensional datasets where exact structure isn't needed, such as preprocessing for clustering or classification. A gotcha is that while distances are preserved probabilistically, the method doesn't capture variance like PCA; it's best for scenarios prioritizing speed over interpretability. For the example of reducing from 1,000 to 100 dimensions, this could compress data while retaining geometric relationships, but performance depends on the dataset's intrinsic structure. If $d$ is too small, distortion increases, so always verify with the JL lemma's bounds.

Let’s break down why $\mathbb{E}[\|\mathbf{P}\mathbf{x}\|^2] = \|\mathbf{x}\|^2$ when $\mathbf{P}$ is a random projection matrix with entries drawn from $\mathcal{N}(0, 1/\sqrt{d})$.

#### Step-by-step Explanation

Let:
- $\mathbf{x} \in \mathbb{R}^n$ (original vector)
- $\mathbf{P} \in \mathbb{R}^{d \times n}$ (random projection matrix)
- Each $P_{ij} \sim \mathcal{N}(0, 1/d)$ (after scaling by $1/\sqrt{d}$)

##### 1. Compute the Projected Vector

The $i$-th component of the projected vector $\mathbf{y} = \mathbf{P}\mathbf{x}$ is:
$$
y_i = \sum_{j=1}^n P_{ij} x_j
$$

##### 2. Compute the Squared Norm

The squared norm is:
$$
\|\mathbf{y}\|^2 = \sum_{i=1}^d y_i^2
$$

##### 3. Compute the Expected Value

We want:
$$
\mathbb{E}[\|\mathbf{P}\mathbf{x}\|^2] = \mathbb{E}\left[\sum_{i=1}^d \left(\sum_{j=1}^n P_{ij} x_j\right)^2\right]
$$

Expand the square:
$$
= \sum_{i=1}^d \mathbb{E}\left[\sum_{j=1}^n \sum_{k=1}^n P_{ij} P_{ik} x_j x_k\right]
$$

Since $P_{ij}$ are independent and have mean $0$:
- If $j \neq k$, $\mathbb{E}[P_{ij} P_{ik}] = 0$
- If $j = k$, $\mathbb{E}[P_{ij}^2] = 1/d$

So:
$$
= \sum_{i=1}^d \sum_{j=1}^n x_j^2 \mathbb{E}[P_{ij}^2]
= \sum_{i=1}^d \sum_{j=1}^n x_j^2 \frac{1}{d}
$$

Sum over $i$:
$$
= \sum_{j=1}^n x_j^2 \left(\sum_{i=1}^d \frac{1}{d}\right)
= \sum_{j=1}^n x_j^2 \cdot 1
= \|\mathbf{x}\|^2
$$

#### Summary

The scaling by $1/\sqrt{d}$ ensures that the expected squared norm of the projected vector equals the original squared norm:
$$
\mathbb{E}[\|\mathbf{P}\mathbf{x}\|^2] = \|\mathbf{x}\|^2
$$

This is why random projection approximately preserves distances and norms, which is crucial for dimensionality reduction.

## Using Scikit-Learn's Random Projection (GaussianRandomProjection)

This section demonstrates how to use Scikit-Learn's `GaussianRandomProjection` to reduce the dimensionality of the synthetic high-dimensional dataset `X` (created earlier). By specifying an error tolerance `eps=ε` (0.1), the transformer automatically determines the target dimension to preserve pairwise distances with high probability, according to the Johnson-Lindenstrauss lemma. Using `fit_transform()`, the data is projected onto a lower-dimensional subspace. The resulting `X_reduced` closely matches the properties of the manual random projection, but with a simpler and more robust implementation. This approach is efficient and well-suited for large datasets where traditional PCA may be too slow or memory-intensive.

````python
from sklearn.random_projection import GaussianRandomProjection

gaussian_rnd_proj = GaussianRandomProjection(eps=ε, random_state=42)
X_reduced = gaussian_rnd_proj.fit_transform(X)  # same result as above
````

**Warning**, the following cell may take several minutes to run:

````python
components_pinv = np.linalg.pinv(gaussian_rnd_proj.components_)
X_recovered = X_reduced @ components_pinv.T
````

where `gaussian_rnd_proj.components_` contains the random projection matrix used for the transformation.

Alternatively, by setting `compute_inverse_components=True` for `GaussianRandomProjection`, you can later reconstruct an approximation of the original data using `inverse_transform()`. This allows you to evaluate the quality of the random projection by comparing the reconstructed data to the original dataset. The inverse transformation uses the pseudo-inverse of the random projection matrix to map the reduced data back to the original space, providing a way to visualize and assess how well the random projection preserved the structure of the data. 

````python
gaussian_rnd_proj_with_inverse = GaussianRandomProjection(eps=ε, random_state=42, compute_inverse_components=True)
X_reduced_2 = gaussian_rnd_proj_with_inverse.fit_transform(X)
X_recovered_2 = gaussian_rnd_proj_with_inverse.inverse_transform(X_reduced_2)

np.allclose(X_recovered, X_recovered_2)
````

# LLE (Locally Linear Embedding)

**LLE (Locally Linear Embedding)** is a nonlinear dimensionality reduction technique that seeks to preserve the local structure of data points in a high-dimensional space when mapping them to a lower-dimensional space. It does this by assuming that each data point and its neighbors lie on or close to a locally linear patch of the manifold. LLE computes weights that best reconstruct each data point from its neighbors in the original space, and then finds a lower-dimensional embedding that preserves these weights. This method is particularly effective for unfolding complex manifolds, such as the Swiss roll, while maintaining the relationships between nearby points.

````python
from sklearn.datasets import make_swiss_roll
from sklearn.manifold import LocallyLinearEmbedding

X_swiss, t = make_swiss_roll(n_samples=1000, noise=0.2, random_state=42)
lle = LocallyLinearEmbedding(n_components=2, n_neighbors=10, random_state=42)
X_unrolled = lle.fit_transform(X_swiss)
````

This code applies Locally Linear Embedding (LLE) to the Swiss roll dataset `X_swiss` for nonlinear dimensionality reduction:

- `lle = LocallyLinearEmbedding(n_components=2, n_neighbors=10, random_state=42)`: Creates an LLE instance to reduce to 2 dimensions, using 10 nearest neighbors to model local linearity, with a fixed random state for reproducibility.
- `X_unrolled = lle.fit_transform(X_swiss)`: Fits the LLE model to `X_swiss` and transforms it into the lower-dimensional space `X_unrolled`, preserving local relationships by reconstructing each point from its neighbors. This unfolds the manifold while maintaining proximity in the original space.

### How LLE Works

Locally Linear Embedding (LLE) is a nonlinear dimensionality reduction technique that preserves the local geometry of data by **modeling each point as a linear combination of its nearest neighbors**. This approach excels at unfolding complex manifolds, such as the Swiss roll, where linear methods like PCA fail due to curvature.

#### Core Idea and Optimization
For each training instance $\mathbf{x}^{(i)}$, LLE identifies its $k$-nearest neighbors and computes weights $w_{i,j}$ that best reconstruct $\mathbf{x}^{(i)}$ as a weighted sum of these neighbors. The weights are found by minimizing the squared reconstruction error:

$$
\mathbf{\widehat{W}}=\operatorname*{argmin}_{\mathbf{W}} \sum_{i=1}^m \left\| \mathbf{x}^{(i)} - \sum_{j=1}^m w_{i,j} \mathbf{x}^{(j)} \right\|^2
$$

subject to the constraints:
- $w_{i,j} = 0$ if $\mathbf{x}^{(j)}$ is not a neighbor of $\mathbf{x}^{(i)}$ (ensures locality); $w_{i,j}$ represents the weight of neighbor $j$ for point $i$.
- $\sum_{j=1}^m w_{i,j} = 1$ for each $i$ (normalizes weights for each point).

This step captures local linear relationships, forming a weight matrix $\mathbf{\widehat{W}}$ that encodes neighborhood structure.

#### Two-Step Process
1. **Modeling Local Relationships**: For each $\mathbf{x}^{(i)}$, find $k$ neighbors and solve the constrained optimization (Equation 8-4) to get weights. This preserves local geometry without assuming global linearity.
   
2. **Mapping to Lower Dimensions**: Embed into $d < n$ dimensions by minimizing the cost where low-dimensional points $\mathbf{z}^{(i)}$ preserve the same weighted combinations (Equation 8-5). The matrix $\mathbf{Z}$ contains the reduced representations.
$$\mathbf{\widehat{Z}}=\operatorname*{argmin}_{\mathbf{Z}} \sum_{i=1}^m \left\| \mathbf{z}^{(i)} - \sum_{j=1}^m w_{i,j} \mathbf{z}^{(j)} \right\|^2$$
This is solved via eigenvalue decomposition of a sparse matrix derived from $\mathbf{\widehat{W}}$, yielding the low-dimensional embedding.
- The eigenvalues and eigenvectors are computed from the matrix $\left(\mathbf{I} - \mathbf{W}\right)^\top \left(\mathbf{I} - \mathbf{W}\right)$, where $\mathbf{I}$ is the identity matrix. The bottom $d+1$ eigenvectors (excluding the smallest eigenvalue corresponding to the trivial solution) provide the coordinates in the reduced space.

#### Intuition and Computational Complexity
LLE assumes local linearity on manifolds, preserving distances in neighborhoods to unfold global structure. It's more complex than PCA, with high computational cost ($O(dm^2)$ for large $m$), and sensitive to noise or varying densities. However, it handles nonlinear manifolds better than linear projections.

#### Comparison to PCA
Unlike PCA, which projects onto orthonormal components (Equation 8-1) to maximize variance (e.g., reducing MNIST from 784D to 154D for 95% variance), LLE prioritizes local weights for manifold unfolding. PCA is efficient for linear data but distorts curved structures like the Swiss roll, where LLE excels by preserving intrinsic geometry.

In summary, LLE learns local relationships first, then embeds them into lower dimensions, making it ideal for nonlinear visualization or preprocessing complex datasets. For linear cases, PCA is faster; for manifolds, LLE or t-SNE may be preferable. If needed, combine PCA for initial reduction before LLE to balance speed and accuracy.

This highlights LLE's ability to capture nonlinear manifold structures that linear methods like PCA cannot. Together, these visualizations emphasize LLE's effectiveness for manifold learning on datasets like the Swiss roll, where preserving local distances leads to a meaningful low-dimensional embedding.

# Other Dimensionality Reduction Techniques

Beyond PCA and LLE, several other dimensionality reduction methods are widely used, each suited to different scenarios:

- **MDS (Multidimensional Scaling)**: Focuses on preserving the *pairwise distances* between points when mapping data to a lower-dimensional space. MDS is useful for visualizing similarities but may not handle complex nonlinear structures as well as other techniques.

- **Isomap**: An extension of Multidimensional Scaling (MDS) that uses *geodesic distances* to better capture the underlying manifold structure. Isomap is effective for "unfolding" nonlinear manifolds like the Swiss roll, but its performance may degrade with noisy or unevenly distributed data.
    - The geodesic distances are computed by constructing a neighborhood graph (using k-nearest neighbors or ε-radius) and applying Dijkstra's or Floyd-Warshall algorithm to find shortest paths between all pairs of points.

- **t-SNE (t-Distributed Stochastic Neighbor Embedding)**: A nonlinear algorithm designed for visualizing high-dimensional data in 2D or 3D. It *preserves local relationships and reveals clusters*, making it popular for exploratory analysis. However, it can be slow and sensitive to parameter choices.

- **LDA (Linear Discriminant Analysis)**: A supervised approach that projects data to *maximize class separation*. LDA is valuable for classification tasks, provided that class labels are available and the data meets certain distributional assumptions.

Note: I added `normalized_stress=False` below to avoid a warning about the fact that the default value for that hyperparameter will change in the future. Please see the [documentation](https://scikit-learn.org/stable/modules/generated/sklearn.manifold.MDS.html) for more details.

# Extra Material – Kernel PCA

Kernel PCA is an extension of traditional PCA that allows for nonlinear dimensionality reduction by using kernel methods. Instead of performing PCA in the original feature space, Kernel PCA first maps the data into a higher-dimensional space using a kernel function (such as the RBF or polynomial kernel) and then performs PCA in that space. This enables Kernel PCA to capture complex, nonlinear relationships in the data that standard PCA cannot. It is particularly useful for datasets where the underlying structure is not linearly separable, such as the Swiss roll dataset. Here's an example of how to use Kernel PCA with Scikit-Learn:

````python
from sklearn.decomposition import KernelPCA

rbf_pca = KernelPCA(n_components=2, kernel="rbf", gamma=0.04, random_state=42)
X_reduced = rbf_pca.fit_transform(X_swiss)
````

- n_components: int — number of projection dimensions to keep (here 2).  
- kernel: str — kernel function used to map data into a higher-dimensional space (e.g., "rbf" = radial basis function).  
- gamma: float — kernel coefficient for "rbf", "poly", and "sigmoid" kernels (controls the influence scale; e.g., 0.04 for the RBF kernel).

````python
lin_pca = KernelPCA(kernel="linear")
rbf_pca = KernelPCA(kernel="rbf", gamma=0.002)
sig_pca = KernelPCA(kernel="sigmoid", gamma=0.002, coef0=1)

kernel_pcas = ((lin_pca, "Linear kernel"),
               (rbf_pca, rf"RBF kernel, $\gamma={rbf_pca.gamma}$"),
               (sig_pca, rf"Sigmoid kernel, $\gamma={sig_pca.gamma}, r={sig_pca.coef0}$"))

plt.figure(figsize=(11, 3.5))
for idx, (kpca, title) in enumerate(kernel_pcas):
    kpca.n_components = 2
    kpca.random_state = 42
    X_reduced = kpca.fit_transform(X_swiss)

    plt.subplot(1, 3, idx + 1)
    plt.title(title)
    plt.scatter(X_reduced[:, 0], X_reduced[:, 1], c=t, cmap=darker_hot)
    plt.xlabel("$z_1$")
    if idx == 0:
        plt.ylabel("$z_2$", rotation=0)
    plt.grid()

plt.show()
````

## The Core Problem: Non-Linear Structures in Data

The Swiss roll dataset represents a fundamental challenge in machine learning: **data that is intrinsically low-dimensional but embedded non-linearly in higher-dimensional space**. While the dataset lives in 3D space, its true underlying structure is just 2D—a rolled-up sheet. Standard linear dimensionality reduction techniques fail to recognize and exploit this structure.

## Why Standard PCA Fails

Standard PCA, which uses a **linear kernel**, cannot "unroll" the Swiss roll because it only finds linear relationships. When you apply PCA to the Swiss roll:

1. **The mathematics remain linear**: PCA computes principal components—orthogonal directions of maximum variance—but these directions are straight lines in the original space.
2. **The result is squashed, not unrolled**: Rather than unfolding the manifold into a clean 2D rectangle, linear PCA simply projects all the curled layers onto a 2D plane. The points remain clustered and overlapped.
3. **Local structure is lost**: Nearby points on the manifold (which are actually close along the surface) may be far apart in the projection, while distant points appear close.

This is why the left plot shows a tight, curved ring-like pattern—the roll's curvature is preserved rather than eliminated.

## The Power of Non-Linear Kernels

Kernel PCA solves this by using the **kernel trick**, which implicitly maps data into a higher-dimensional feature space without explicitly computing those coordinates. This allows the algorithm to find non-linear patterns that would appear linear in the transformed space.

### RBF Kernel (Middle Plot)

The **Radial Basis Function (RBF) kernel** with `γ=0.002` creates a local similarity measure:

- **What it does**: Maps nearby points closer together and distant points further apart, based on local neighborhood structure.
- **Result on Swiss roll**: The data becomes more spread out and less tightly curled. The color gradient (yellow to red) is still visible, indicating that the original ordering along the roll is partially preserved.
- **Limitation**: With `γ=0.002` (relatively large), the kernel operates at a broad scale. A smaller `γ` value would create more local, fine-grained neighborhoods and might unroll the manifold more completely.

The RBF kernel is particularly effective for capturing **local non-linear structures** because it focuses on distances between nearby points rather than global patterns.

### Sigmoid Kernel (Right Plot)

The **sigmoid kernel** with `γ=0.002` and `coef0=1` behaves somewhat like a neural network activation function:

$$K(x, y) = \tanh(\gamma \langle x, y \rangle + \text{coef0})$$

- **What it does**: Maps data through a non-linear, smooth function that resembles neural network activations.
- **Result on Swiss roll**: Creates a dispersed cloud-like pattern with a darker center. The structure is more diffuse than the RBF kernel, suggesting a different way of measuring similarity.
- **Behavior**: The sigmoid kernel can be less stable and more sensitive to parameter tuning than RBF. Its performance varies significantly depending on both `γ` and `coef0`.

## Key Insights from This Visualization

1. **Linear methods have fundamental limits**: No amount of parameter tuning will make standard PCA unroll the Swiss roll. The linear constraint is insurmountable.

2. **Kernels matter**: Different kernels capture different aspects of non-linear structure. RBF is generally better for manifold learning tasks, while sigmoid is more specialized.

3. **Parameter tuning is crucial**: The results shown here use fixed `γ` values. Smaller `γ` would create tighter, more local neighborhoods—potentially unrolling the manifold more effectively.

4. **Kernel PCA is not a complete solution**: Even with non-linear kernels, Kernel PCA may not perfectly unroll complex manifolds. For the Swiss roll specifically, other algorithms like **LLE (Locally Linear Embedding)** or **Isomap** are often superior because they explicitly model the manifold's geodesic structure.

## When to Use Each Kernel

- **Linear kernel**: Use as a baseline; equivalent to standard PCA. Best for datasets with primarily linear relationships.
- **RBF kernel**: Default choice for non-linear dimensionality reduction. Excellent for capturing local structure and general-purpose manifold learning.
- **Polynomial kernel**: Good for finding polynomial relationships; useful when you suspect smooth, polynomial structure in the data.
- **Sigmoid kernel**: Specialized use cases; less commonly recommended due to parameter sensitivity and variable performance.

## Practical Takeaway

This example beautifully illustrates a central theme in machine learning: **the choice of algorithm and parameters fundamentally shapes what patterns you can discover**. Linear methods like PCA see only linear patterns; non-linear kernels reveal hidden structures invisible to linear approaches. However, even with powerful non-linear tools, understanding your data's intrinsic geometry remains essential for choosing the right technique.

### Explanation of CPU Time and Wall Time

The `%time` magic command in Jupyter measures the execution time of the code cell. It reports two key metrics:

- **Wall time**: This is the total elapsed real time from the start to the end of the execution, including any time spent waiting for I/O operations, other processes, or system overhead. It's the "clock on the wall" time you would experience waiting for the code to finish. In the output above, this is shown as "Wall time: X.XX s".

- **CPU time**: This measures the actual time the CPU(s) spent executing the code. It can be less than wall time if the process was idle (e.g., waiting for disk or network), or greater than wall time if the code uses multiple CPU cores (parallel execution). In the output, it's shown as "CPU times: user X.XX s, sys: Y.YY s, total: Z.ZZ s", where:
    - `user`: Time spent in user-mode code (your script).
    - `sys`: Time spent in system/kernel-mode code (e.g., I/O operations).
    - `total`: Sum of user and sys time.

For the Random Forest training, wall time indicates how long you wait, while CPU time shows the computational intensity. If CPU time is much less than wall time, there might be I/O bottlenecks; if higher, the model is CPU-bound and could benefit from parallelization.

Let's create a `plot_digits()` function that will draw a scatterplot (similar to the above scatterplots) plus write colored digits, with a minimum distance guaranteed between these digits. If the digit images are provided, they are plotted instead. This implementation was inspired from one of Scikit-Learn's excellent examples ([plot_lle_digits](https://scikit-learn.org/stable/auto_examples/manifold/plot_lle_digits.html), based on a different digit dataset).

````python
from sklearn.preprocessing import MinMaxScaler
from matplotlib.offsetbox import AnnotationBbox, OffsetImage

def plot_digits(X, y, min_distance=0.04, images=None, figsize=(13, 10)):
    # Let's scale the input features so that they range from 0 to 1
    X_normalized = MinMaxScaler().fit_transform(X)
    # Now we create the list of coordinates of the digits plotted so far.
    # We pretend that one is already plotted far away at the start, to
    # avoid `if` statements in the loop below
    neighbors = np.array([[10., 10.]])
    # The rest should be self-explanatory
    plt.figure(figsize=figsize)
    cmap = plt.cm.jet
    digits = np.unique(y)
    for digit in digits:
        plt.scatter(X_normalized[y == digit, 0], X_normalized[y == digit, 1],
                    c=[cmap(float(digit) / 9)], alpha=0.5)
    plt.axis("off")
    ax = plt.gca()  # get current axes
    for index, image_coord in enumerate(X_normalized):
        closest_distance = np.linalg.norm(neighbors - image_coord, axis=1).min()
        if closest_distance > min_distance:
            neighbors = np.r_[neighbors, [image_coord]]
            if images is None:
                plt.text(image_coord[0], image_coord[1], str(int(y[index])),
                         color=cmap(float(y[index]) / 9),
                         fontdict={"weight": "bold", "size": 16})
            else:
                image = images[index].reshape(28, 28)
                imagebox = AnnotationBbox(OffsetImage(image, cmap="binary"),
                                          image_coord)
                ax.add_artist(imagebox)
````
